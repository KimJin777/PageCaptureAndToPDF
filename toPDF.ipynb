```python
import io
import os
import pytesseract
from pypdf import PdfReader, PdfWriter  # [수정] 안정적인 PdfWriter로 변경

# ==========================================
# [설정] Tesseract OCR 프로그램 설치 경로 지정
# ==========================================
TESSERACT_PATH = r"C:\Program Files\Tesseract-OCR\tesseract.exe"
pytesseract.pytesseract.tesseract_cmd = TESSERACT_PATH


def convert_images_to_final_pdf(image_folder, output_pdf_name):
    """pypdf.PdfWriter를 사용하여 버전 제한 없이 안정적으로 PDF를 누적 병합하는 함수"""

    # 1. 폴더 내 이미지 파일 목록 정렬하여 가져오기
    valid_extensions = (".jpg", ".jpeg", ".png")
    image_files = [
        f
        for f in os.listdir(image_folder)
        if f.lower().endswith(valid_extensions)
    ]
    image_files.sort()

    if not image_files:
        print(f"❌ '{image_folder}' 폴더 내에 이미지 파일이 없습니다.")
        return

    print(f"📚 총 {len(image_files)}개의 이미지를 변환합니다.")
    print("⏳ 대량의 OCR 분석 및 PDF 누적 병합을 시작합니다...\n")

    # [수정] 페이지를 순서대로 추가해 나갈 고정 Writer 객체 생성
    writer = PdfWriter()
    success_count = 0

    for idx, filename in enumerate(image_files, start=1):
        img_path = os.path.join(image_folder, filename)
        print(
            f"[{idx}/{len(image_files)}] '{filename}' OCR 텍스트 분석 중...",
            end="",
            flush=True,
        )

        try:
            # 2. pytesseract로 텍스트 레이어가 포함된 개별 PDF 바이너리 데이터 생성
            pdf_data = pytesseract.image_to_pdf_or_hocr(img_path, lang="kor+eng")

            # 3. 메모리 상의 PDF 바이트 데이터를 PdfReader로 읽어서 writer에 한 장씩 누적
            pdf_stream = io.BytesIO(pdf_data)
            reader = PdfReader(pdf_stream)

            # 생성된 PDF의 첫 번째 페이지(page[0])를 최종 문서에 추가
            writer.add_page(reader.pages[0])

            success_count += 1
            print(" 성공 📥 누적됨")

        except Exception as e:
            print(f" 실패 (오류: {e})")

    # 4. 결합된 최종 PDF 저장
    if success_count > 0:
        print(f"\n💾 총 {success_count}개 페이지 최종 병합 및 파일 저장 중...")
        try:
            with open(output_pdf_name, "wb") as f_out:
                writer.write(f_out)
            writer.close()

            print("\n==================================================")
            print("🎉 모든 페이지가 완벽하게 누적된 PDF가 완성되었습니다!")
            print(f"📁 저장된 파일: {os.path.abspath(output_pdf_name)}")
            print("==================================================")
        except Exception as e:
            print(f"❌ 파일 저장 중 오류 발생: {e}")
    else:
        print("❌ 변환에 성공한 페이지가 없어 PDF 파일이 생성되지 않았습니다.")




if __name__ == "__main__":
    # 사용자에게 실행 시 대상 폴더와 출력 파일명을 입력받도록 변경
    default_folder = "./captured_YYYYMMDD_HHMMSS"
    folder_input = input(f"대상 이미지 폴더를 입력하세요 (상대경로 또는 절대경로, 기본: {default_folder}): ").strip()
    TARGET_FOLDER = folder_input if folder_input else default_folder

    default_output = "output.pdf"
    output_input = input(f"결과 PDF 파일명을 입력하세요 (기본: {default_output}): ").strip()
    OUTPUT_FILE = output_input if output_input else default_output

    print(f"선택된 폴더: {TARGET_FOLDER}")
    print(f"출력 파일: {OUTPUT_FILE}")

    convert_images_to_final_pdf(TARGET_FOLDER, OUTPUT_FILE)
```
