```python
import datetime
import os
import time
from mss import mss
from PIL import Image
import pyautogui


def capture_all_pages_with_clean_name(total_pages=56):
    """오른쪽 화살표를 누르며 지정된 페이지 수만큼 001, 002 순서로 캡처하는 함수"""

    with mss() as sct:
        if len(sct.monitors) < 3:
            print(
                "인식된 모니터가 2개 미만입니다. 듀얼 모니터 연결 상태를 확인하세요."
            )
            return

        right_monitor = sct.monitors[1]

        print("==================================================")
        print("  📢 5초 후에 자동 캡처 및 페이지 넘김을 시작합니다.  ")
        print("  👉 지금 오른쪽 모니터의 웹페이지 화면을 한번 클릭해 주세요!  ")
        print("==================================================")

        # 5초 대기 시간 동안 오른쪽 웹페이지 포커스 활성화
        for i in range(5, 0, -1):
            print(f"{i}초 전...")
            time.sleep(1)

        print("\n🚀 캡처를 시작합니다. 마우스와 키보드를 움직이지 마세요.\n")

        # [구분용] 실행할 때마다 고유한 폴더 생성 (예: captured_20260520_094500)
        folder_time = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
        output_dir = f"captured_{folder_time}"
        os.makedirs(output_dir, exist_ok=True)

        for page in range(1, total_pages + 1):
            print(
                f"[{page} / {total_pages}] 페이지 캡처 중...", end="", flush=True
            )

            # 1. 오른쪽 화면 캡처
            sct_img = sct.grab(right_monitor)

            # 2. [수정] 파일명을 요청하신 대로 순수한 번호(001.jpg, 002.jpg)로만 지정
            filename = os.path.join(output_dir, f"{page:03d}.jpg")
            img = Image.frombytes(
                "RGB", sct_img.size, sct_img.bgra, "raw", "BGRX"
            )
            img.save(filename, "JPEG")

            print(" 완료 ➡️ 다음 페이지")

            # 3. 다음 페이지로 이동 처리
            if page < total_pages:
                pyautogui.press("right")
                time.sleep(1)  # 웹페이지 로딩 대기 시간 (필요시 늘려주세요)

        print("\n==================================================")
        print(f"🎉 총 {total_pages}페이지 캡처 작업이 완료되었습니다!")
        print(f"📁 저장된 폴더 경로: {os.path.abspath(output_dir)}")
        print("==================================================")


if __name__ == "__main__":
    # 총 367페이지 실행
    page_number = N # 페이지 수 지정
    capture_all_pages_with_clean_name(total_pages=page_number)
)
```
